```python
import pandas as pd
```

## Import and prepare the data table

```python
df = pd.read_excel('city_data.xlsx', skiprows=4)
```

```python
df = df.iloc[:, 2:].set_index('Unnamed: 2')
```

## Construct an empty grid of the cities

```python
# extract city names
cities = df.index.tolist()
```

```python
grid = pd.DataFrame(index=cities, columns=cities)
```

## Fill in grid using the distance formula

```python
df.head()
```

```python
for x in cities:
    for y in cities:
        lat1 = df.loc[x, 'Lat']
        long1 = df.loc[x, 'Long']
        
        lat2 = df.loc[y, 'Lat']
        long2 = df.loc[y, 'Long']
        
        distance = round(69 * ((lat1 - lat2)**2 + (long1 - long2)**2)**0.5)
        
        grid.loc[x, y] = distance
```
