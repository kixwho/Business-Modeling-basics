```python
import pandas as pd
import numpy as np
```

### Input assumptions:
                    annual demand = 65000 - 9000*price
                    fixed cost = 45000

### Model how annual profit changes as price and unit cost vary

```python
fixed_cost = 45000
```

```python
prices = np.arange(2.5, 5.01, 0.25)
unit_costs = np.arange(0.3, 0.61, 0.05)
```

```python
sensitivity_table = pd.DataFrame(index=prices, columns=unit_costs)
```

```python
for x in prices:
    for y in unit_costs:
        demand = 65000 - 9000 * x
        revenue = x * demand
        variable_cost = y * demand
        profit = revenue - fixed_cost - variable_cost
        
        sensitivity_table.loc[x,y] = profit
```

```python
sensitivity_table = sensitivity_table.rename_axis('price', axis=0).rename_axis('unit cost', axis=1)
sensitivity_table
```

```python
styled_table = (
    sensitivity_table
    .style
    .highlight_max(axis=0, props='font-weight:bold; background-color:yellow') # highlight the profit-maximizing price
    .format('{:.2f}')
    .format_index('{:.2f}', axis=0)
    .format_index('{:.2f}', axis=1)
    .set_caption('Profit as a function of Price and Unit Cost')
    .set_table_styles([
        {
            'selector': 'th.col_heading',
            'props': 'background-color: #CDE6D0'
        },
        {
            'selector': 'th.row_heading',
            'props': 'background-color: #CDE6D0'
        },
        {
            'selector': 'caption',
            'props': 'font-size: 18px'
        }
    ])
)
```

```python
display(styled_table)
```

### Conclusion

### As the unit cost increases, the profit-maximizing price increases as some of the cost increase is passed on to customers
